```python
titanic=titanic.iloc[:,2:]
```

```python
titanic.sample(5)
```

```python
titanic.info()
```

```python
##Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(titanic.drop('Fare', axis=1),
                                                   titanic['Fare'],
                                                   test_size=0.3,
                                                   random_state=0)
X_train.shape, X_test.shape
```

```python
##Filtering numeric data
X_train = X_train.select_dtypes(include=['int64', 'float64'])
X_test = X_test.select_dtypes(include=['int64', 'float64'])
```

```python
##StandardScaler
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# fit the scaler to the train set, it will learn the parameters
scaler.fit(X_train)

# transform train and test sets
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler.mean_
```

```python
titanic.info()
```

```python
X_train
```

```python
X_train_scaled
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
np.round(X_train.describe(), 1)
```

```python
##Effect of Scaling
# Assuming X_train and X_train_scaled are defined DataFrames
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# Plot Before Scaling
ax1.scatter(X_train['Age'], X_train['Pclass']) # Added quotes
ax1.set_title('Before Scaling')               # Added quotes

# Plot After Scaling
ax2.scatter(X_train_scaled['Age'], X_train_scaled['Pclass'], color='yellow') # Fixed typo ('Agw'->'Age'), fixed keyword ('colors'->'color')
ax2.set_title('After Scaling')                # Added quotes

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))
##before scaling
ax1.set_title('Befor Scaling')
sns.kdeplot(X_train['Age'], ax=ax1)
sns.kdeplot(X_train['Pclass'], ax=ax1)

##after scaling
ax2.set_title('After Scaling')
sns.kdeplot(X_train_scaled['Age'], ax=ax2)
sns.kdeplot(X_train['Pclass'], ax=ax2)

plt
```

```python
##comparision of distribution
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))
##before scaling
ax1.set_title('Age distribution before scaling')
sns.kdeplot(X_train['Age'], ax=ax1)

##after scaling
ax2.set_title('Age distribution after scaling')
sns.kdeplot(X_train_scaled['Age'], ax=ax2)
plt.show()          
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))
##before scaling
ax1.set_title('Pclass distribution before scaling')
sns.kdeplot(X_train['Pclass'], ax=ax1)

##after scaling
ax2.set_title('Pclass distribution after scaling')
sns.kdeplot(X_train_scaled['Pclass'], ax=ax2)
plt.show()          
```
