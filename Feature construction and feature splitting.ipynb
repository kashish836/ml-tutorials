```python
import numpy as np 
import pandas as pd
import seaborn as sns 

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('train.csv')[['Age', 'Pclass', 'SibSp', 'Parch', 'Survived']]
```

```python
df.head()
```

```python
"""permanently modifies your original DataFrame by removing all rows that contain missing values (NaN or None) ------>>>>>>"""
```

```python
df.dropna(inplace=True)
```

```python
df.head()
```

```python
X = df.iloc[:,0:4]
Y = df.iloc[:,-1]
```

```python
X.head()
```

```python
Y.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,Y,scoring='accuracy',cv=20))
##This code calculates the average accuracy of a Logistic Regression model after testing it across 20 different splits of your data to ensure the performance score is reliable.
```

```python
##APPLYING FEATURE CONSTRUCTION
```

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
```

```python
X.head(10)
```

```python
def myfunc(num):
    if num == 1:
        #alone
        return 0
    elif num >1 and num <=4:
        #small family
        return 1
    else:
        #large family
        return 2
```

```python
myfunc(7)
```

```python
X['Family_type'] = X['Family_size'].apply(myfunc)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,Y,scoring='accuracy',cv=20))
```
